```python
import numpy as np
import matplotlib.pyplot as plt
def f(x):
    return (x+3)**2
def df(x):
    return 2*(x+3)
x=2
learning_rate=0.1
epochs=30
for i in range(epochs):
    grad=df(x)
    x=x-learning_rate*grad
    print('iteration : ',i+1,' x: ',x,' f(x) ',f(x))
print('local minimum occurs at x= ',x )
print('minimum value of function = ',f(x))
    
```

```python
x=np.linspace(-10,5,200)
y=f(x)
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.show()
```
