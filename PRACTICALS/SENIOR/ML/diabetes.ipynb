```python
import pandas as pd 
from sklearn.metrics import confusion_matrix,accuracy_score,recall_score,precision_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
data=pd.read_csv('diabetes.csv')
data.head()
```

```python
X=data.drop(columns=['Outcome'])
y=data['Outcome']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train,y_train)
y_pred_knn=knn.predict(X_test)
```

```python
print('accuracy ',accuracy_score(y_pred_knn,y_test))
print('confusion matrix ',confusion_matrix(y_pred_knn,y_test))
print('recall ',recall_score(y_pred_knn,y_test))
print('precision ',precision_score(y_pred_knn,y_test))
print('error rate ',1-accuracy_score(y_pred_knn,y_test))
```
